## Command Line Interface

Our package features a Python command-line interface (CLI) script called "dcTMD". 

The script loads the constraint force files, generates a work/force set, and calculates the mean work, dissipated work, free energy, and friction factor. The output is saved as a .npz and .dat file, and the work/force set can be saved separately using the "--save_dataset" option. Furthermore, with the --plot option an overview plot of the results in generated.

The CLI also provides a verbose mode for debugging and error reporting.

To get an overview over all modules simply run the following command in the shell.

```python
%%bash
python -m dcTMD --help
```

```python
# Usage example:
python -m dcTMD -f '*.xvg' -o path/to/output/name  -T 290.5 -v 0.001 --verbose --plot --save_dataset
```

#### A note on input file paths

The `-f` option accepts either a glob pattern or a text file that lists one force trajectory per line. The two forms resolve paths as follows:

A glob pattern is expanded relative to the current working directory, exactly as in the shell.

A list file is read line by line and the listed paths are passed through unchanged. They are therefore opened relative to the current working directory, not relative to the location of the list file.

```python
%%bash
# The list file exists, but the paths inside have the form 
# 'testdata/t_middle_01_pullf.xvg'
# Thus, the following command fails when run from the project root.
pwd
python -m dcTMD -f ../../tests/testdata/pullf_filenames.dat -o /tmp/dcTMD_fails -T 290.15 -vel 0.001
```

When `dcTMD` is invoked from the `tests/` directory the run is successfull.

```python
%%bash
cd ../../tests
pwd
python -m dcTMD -f testdata/pullf_filenames.dat \
    -o /tmp/dcTMD_list -T 290.15 -vel 0.001
```
